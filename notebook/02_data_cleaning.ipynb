```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("../data/city_day.csv")
data.head()
```

```python
data.info()
data.isnull().sum()
```

```python
# Drop rows with missing target or major features
data.dropna(subset=['AQI'], inplace=True)

# Fill smaller gaps with median or mean
data.fillna(data.median(numeric_only=True), inplace=True)
```

```python
data['AQI'] = pd.to_numeric(data['AQI'], errors='coerce')
```

```python
Q1 = data['AQI'].quantile(0.25)
Q3 = data['AQI'].quantile(0.75)
IQR = Q3 - Q1
data = data[~((data['AQI'] < (Q1 - 1.5 * IQR)) | (data['AQI'] > (Q3 + 1.5 * IQR)))]
```

```python
data.to_csv('data/cleaned_aqi_data.csv', index=False)
```
